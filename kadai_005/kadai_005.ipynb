```python
'''
課題5
台形の面積を求めるプログラム
'''
# 上辺：10cm
traps_top = 10
# 下辺：20cm
traps_bot = 20
# 高さ：5cm
traps_hit = 5

# 公式：台形の面積 =（上辺＋下辺）× 高さ ÷ 2
area = (traps_top + traps_bot) * traps_hit / 2
print(f" 台形の面積は {area} cm²")
```
