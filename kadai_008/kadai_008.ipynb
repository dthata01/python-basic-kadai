```python
'''
課題 ８ 条件分岐のif文で処理を切り替えよう
'''

var = 11

# var が3で割り切れたら 「Fuzz」
if ( var % 3 == 0) & (var % 5 != 0 ):
  print("Fuzz")
# var が5で割り切れたら 「Buzz」
elif ( var % 3 != 0 ) & (var % 5 == 0):
  print("Buzz")
# var が3と5で割り切れたら 「FuzzBuzz」
elif ( var % 3 == 0) & (var % 5 == 0):
  print("FuzzBuzz")
# いずれでもなければ、varを表示
else:
  print(var)
```
