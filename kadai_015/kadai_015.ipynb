```python
'''
課題15 クラスを使いこなそう
'''

# Human クラスの作成
class Human:
  # クラスの初期化
  def __init__(self):
    self.name = ''
    self.age = 0
  # 情報の出力
  def printinfo(self):
    print( self.name )
    print( self.age )

# インスタンスの作成
human = Human()

# 属性に値を代入
human.name = '侍太郎'
human.age = 25

# 情報の表示
human.printinfo()
```
