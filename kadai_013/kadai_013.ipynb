```python
'''
課題13 関数を使いこなそう
'''
# 関数 calc_price 消費税を足した金額を計算する。
def calc_price( price: int, rate: int):
  # 金額 + 消費税分(= 金額 ＊ 消費税 / 100 )を返す
  return price + price * rate / 100

# 計算結果を直接出力
print( calc_price( 110, 10) )
```
