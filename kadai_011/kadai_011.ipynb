```python
'''
課題11 繰り返し処理を使いこなそう
'''
# 配列に値を格納

array = ["水","金","地","火","木","土","天","海","冥"]

# forで配列の繰り返し処理を行います。
for star_name in array:
  print( star_name )

# whileで配列の繰り返し処理を行います。
# カウンタの初期化 リストだから0スタート
cnt = 0
while ( cnt < 9):
  print(array[cnt])
  # カウンタのインクリメント
  cnt = cnt + 1
```
