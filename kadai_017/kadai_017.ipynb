```python
'''
課題17 学習したことを組み合わせてコードを書こう
'''

# Human クラスの作成
class Human:
  def __init__(self, name:str, age:int):
    self.name = name
    self.age = age

  def check_adult(self):
    if ( self.age >= 20 ):
      print("大人である。")
    else:
      print("大人でない。")

# 複数インスタンスをリスト化しての作成、追加
name_list = [Human("侍太郎", 40)]
name_list.append(Human("侍一郎", 20))
name_list.append(Human("侍二郎", 19))
name_list.append(Human("侍花子", 25))
name_list.append(Human("侍梅子", 15))

# リストのすべての要素で関数を実行
# リストの要素をhumanに取り出して関数を実行
for human in name_list:
  human.check_adult()
```
